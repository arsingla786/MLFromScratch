# **Date**

```python
import pandas as pd
import numpy as np

# For reproducibility
np.random.seed(42)

# Generate 200 rows
n = 200

data = {
    "date": pd.to_datetime(
        np.random.choice(
            pd.date_range("2018-01-01", "2019-12-31"),
            size=n
        )
    ),
    "product_id": np.random.randint(1000, 9999, size=n),
    "city_id": np.random.randint(1, 31, size=n),
    "orders": np.random.randint(1, 200, size=n)
}

df = pd.DataFrame(data)

# Sort by date (optional)
df = df.sort_values("date").reset_index(drop=True)

print(df.head())
```

```python
df.info()
```

# **1. Extract year , month, day**

```python
df['date_year'] = df['date'].dt.year
df.head()
```

```python
df['date_month'] = df['date'].dt.month
#use dt.month_name() for name of month
df.head()
```

```python
#day of week
df['date_dow'] = df['date'].dt.dayofweek
df.head()
```

```python
#quarter
df['date_quarter'] = df['date'].dt.quarter
df.head()
```
