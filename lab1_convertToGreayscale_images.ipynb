```python
import cv2
```

```python
img_path = '/content/Gemini_Generated_Image_jb0zzvjb0zzvjb0z.png'
```

```python
img_path = cv2.imread(img_path)
```

```python
greyscale_img = cv2.cvtColor(img_path, cv2.COLOR_BGR2GRAY)
original_img = cv2.cvtColor(img_path, cv2.COLOR_BGR2RGB)
```

```python
print(greyscale_img.min())
print(greyscale_img.max())
print(greyscale_img.mean())
print(greyscale_img.sum())
print(greyscale_img.shape)
print(original_img.shape)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

# Display Original Image
plt.subplot(1, 2, 1)
plt.imshow(original_img)
plt.title('Original Image')
plt.axis('off')

# Display Grayscale Image
plt.subplot(1, 2, 2)
plt.imshow(greyscale_img, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')

plt.show()
```

```python
import cv2
```

```python
img3_path = '/content/pic.jpg'
```

```python
img3 = cv2.imread(img3_path,0)
```

```python
print(img3)
print(img3.shape)
```

```python
img3_mean = img3.mean()
```

```python
print(img3_mean)
```

```python
for i in range(img3.shape[0]):
    for j in range(img3.shape[1]):
        if img3[i, j] >= img3_mean:
            img3[i, j] = 255
        else:
            img3[i, j] = 0
```

```python
print(img3)
```

```python
plt.imshow(img3,cmap = 'gray')
```
