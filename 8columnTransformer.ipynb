# **Column Transformer** -> perform multiple preprocessing on different columns in same block

```python
import seaborn as sns
df = sns.load_dataset('titanic')
```

```python
df.head()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder
transformer = ColumnTransformer(transformers= [
    ('t1',SimpleImputer(),['age']),
    ('t2',OneHotEncoder(drop='first'),['embarked','sex','embark_town','alive']),
    ('t3',OrdinalEncoder(categories = [['Third','Second','First']]),['class'])
],remainder='passthrough')
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest  =  train_test_split(df.drop('survived',axis=1),df['survived'],test_size=0.2,random_state=0)
```

```python
xtrain_transformed = transformer.fit_transform(xtrain)
xtest_transformed = transformer.transform(xtest)
```

```python
import pandas as pd
xtrain_transformed = pd.DataFrame(xtrain_transformed,columns=transformer.get_feature_names_out())
```

```python
xtest_transformed = pd.DataFrame(xtest_transformed,columns=transformer.get_feature_names_out())
```
