```python
from keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
img  = image.load_img('/content/Screenshot 2026-10-17 231528.png',target_size=(200,200))
```

```python
import matplotlib.pyplot as plt
plt.imshow(img,)
```

```python
datagen = ImageDataGenerator(
    rotation_range=40,
    shear_range =  0.2,
    zoom_range = 0.2,
    horizontal_flip = True,
    width_shift_range = 0.2,
    height_shift_range = 0.2
)
```

```python
img = image.img_to_array(img)   #convert to numpy array
```

```python
img.shape
```

```python
type(img)
```

```python
input_batch = img.reshape(1,200,200,3)
```

##this will generate 10 new imagesusing the given image after applying random transformations in datagen

```python
i = 0
for output in datagen.flow(input_batch,batch_size=1,save_to_dir='/content'):
  i = i+1
  if i==10 :
    break
```
